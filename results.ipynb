```python
import pandas as pd
import numpy as np
```

```python
results_df = pd.read_csv('model_comparison_results.csv')
```

```python
results_df[results_df['Modelo'].str.contains('Otimo')]
```

```python
modelos = ['Ridge', 'Lasso', 'Random Forest', 'XGBoost', 'KNN', 'SVR']
results_dict = dict()
```

```python
final_results_df = results_df[~results_df['Modelo'].str.contains('Otimo') & ~results_df['Modelo'].str.contains('Regressao Linear')]
final_results_df = final_results_df.drop(columns=['RMSE Treino', 'RMSE Teste',
       'MAE Treino', 'MAE Teste', 'Tempo (segundos)', 'R2 Treino'])
```

```python
for modelo in modelos:
    modelo_df = final_results_df[final_results_df['Modelo'].str.contains(modelo) & ~final_results_df['Modelo'].str.contains('Otimo')]
    results_dict[modelo] = modelo_df
```
